# Part 1: Social Listening Warm-up
*งานเดี่ยว — ทุกคนต้องทำและส่งของตนเอง ไม่ต้องนำเสนอ (10 คะแนน)*

ใช้ข้อมูลโพสต์สาธารณะจาก **Bluesky**  
https://huggingface.co/datasets/alpindale/two-million-bluesky-posts

ใช้เพียง **1–3 ไฟล์** ก็พอ ไม่จำเป็นต้องโหลดทั้ง dataset

ฟิลด์ที่มี เช่น `text`, `created_at`, `author`, `uri`, `has_images`, `reply_to`, `predicted_language`

## ตอบ 4 คำถามหลัก

1. **คนโพสต์เมื่อไร:** จำนวนโพสต์ ช่วงเวลาที่ข้อมูลครอบคลุม และชั่วโมงที่โพสต์มากที่สุด
2. **ใช้ภาษาอะไร:** รายงานภาษาหลักและสัดส่วน เพื่อบอกว่าถ้าทำคอนเทนต์ควรรองรับภาษาใด
3. **คนพูดถึงอะไร:** หา hashtag หรือคำ/วลีเด่นที่น่าสนใจ และยกตัวอย่างข้อความจริงประกอบ
4. **ข้อมูลนี้บอกอะไรได้และบอกอะไรไม่ได้:** สร้าง visualization 1 รูป แล้วเขียนข้อจำกัด 3–5 บรรทัด

> Part 1 เป็น warm-up ไม่ต้องทำ influencer analysis หรือ reply-network เว้นแต่สนใจทำเพิ่มเอง


```python
# ----------------- Part 1: Load Bluesky data -----------------
from huggingface_hub import HfApi, hf_hub_download
import pandas as pd
import matplotlib.pyplot as plt
import re

REPO_ID = "alpindale/two-million-bluesky-posts"
api = HfApi()
files = api.list_repo_files(REPO_ID, repo_type="dataset")
jsonl_files = [f for f in files if f.endswith(".jsonl") or f.endswith(".jsonl.gz")]
print("JSONL files:", jsonl_files)

DATA_FILE = jsonl_files[0]
local_file = hf_hub_download(repo_id=REPO_ID, filename=DATA_FILE, repo_type="dataset")

def iter_jsonl(path):
    import json
    import gzip
    opener = gzip.open if path.endswith(".gz") else open
    with opener(path, "rt", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                yield json.loads(line)

rows = []
MAX_READ = 50000
for i, row in enumerate(iter_jsonl(local_file)):
    rows.append(row)
    if i + 1 >= MAX_READ:
        break

df_bluesky = pd.DataFrame(rows)
print("Rows:", len(df_bluesky))
print("Columns:", df_bluesky.columns.tolist())
df_bluesky.head()
```

```python
# ----------------- Part 1: Answer the 4 questions -----------------
df = df_bluesky.copy()

df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce", utc=True)
df = df.dropna(subset=["created_at"]).copy()

# Q1: When do people post?
df["hour"] = df["created_at"].dt.hour
hour_count = df["hour"].value_counts().sort_index()
peak_hour = hour_count.idxmax()
print("Q1: ช่วงเวลาที่มีโพสต์มากที่สุด (UTC):", peak_hour, "นาฬิกา")
print("ช่วงข้อมูล:", df["created_at"].min(), "ถึง", df["created_at"].max())

# Q2: What language is used?
if "predicted_language" in df.columns:
    lang = df["predicted_language"].fillna("unknown").value_counts().head(10)
    print("\nQ2: ภาษาที่พบมากที่สุด")
    print(lang)
    print("\nเปอร์เซ็นต์")
    print((lang / len(df) * 100).round(2))
else:
    print("Q2: ไม่พบคอลัมน์ predicted_language ในข้อมูล")

# Q3: What do people talk about? Use hashtags as a simple indicator.
texts = df["text"].fillna("")
hashtags = []
for text in texts:
    hashtags.extend(re.findall(r"(?<!\w)#([A-Za-z0-9_ก-๙]+)", text))

hashtag_count = pd.Series(hashtags).value_counts().head(10)
print("\nQ3: Hashtag ที่พบบ่อย")
print(hashtag_count)

if len(hashtag_count) > 0:
    top_tag = hashtag_count.index[0]
    examples = df[df["text"].str.contains("#" + re.escape(top_tag), case=False, na=False)]["text"].head(3)
    print("\nตัวอย่างโพสต์ที่มี #" + top_tag)
    for x in examples:
        print("-", x)

# Q4: Visualization
plt.figure(figsize=(10, 4))
hour_count.plot(kind="bar")
plt.title("Number of Bluesky Posts by Hour (UTC)")
plt.xlabel("Hour (UTC)")
plt.ylabel("Number of posts")
plt.tight_layout()
plt.show()

print("\nข้อจำกัดของข้อมูล")
limitations = [
    "1. ข้อมูลเป็นโพสต์สาธารณะ จึงอาจไม่แทนผู้ใช้ Bluesky ทั้งหมด",
    "2. การอ่านเพียงบางส่วนของ dataset อาจให้ผลต่างจากข้อมูลทั้งหมด",
    "3. เวลาในข้อมูลเป็น UTC จึงอาจไม่ตรงกับเวลาท้องถิ่นของผู้ใช้",
    "4. ภาษาและ hashtag ขึ้นอยู่กับวิธีตรวจจับและการเก็บข้อมูลของ dataset",
    "5. จำนวนโพสต์หรือ hashtag มาก ไม่ได้แปลว่าผู้ใช้ต้องการสิ่งนั้นเสมอไป"
]
for x in limitations:
    print(x)
```
